```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
### Load the trained model, scaler pickle,onehot
model=load_model('model.h5')

with open('label_encoder_accidental.pkl','rb') as file:
    label_encoder_accidental=pickle.load(file)

with open('label_encoder_transmission.pkl','rb') as file:
    label_encoder_transmission=pickle.load(file)

with open('onehot_encoder_brand.pkl','rb') as file:
    onehot_encoder_brand=pickle.load(file)

with open('onehot_encoder_fuel_type.pkl','rb') as file:
    onehot_encoder_fuel_type=pickle.load(file)

with open('onehot_encoder_model_name.pkl','rb') as file:
    onehot_encoder_model_name=pickle.load(file)

with open('onehot_encoder_model_variant.pkl','rb') as file:
    onehot_encoder_model_variant=pickle.load(file)

with open('onehot_encoder_car_type.pkl','rb') as file:
    onehot_encoder_car_type=pickle.load(file)

with open('onehot_encoder_state.pkl','rb') as file:
    onehot_encoder_state=pickle.load(file)

with open('onehot_encoder_owner.pkl','rb') as file:
    onehot_encoder_owner=pickle.load(file)

with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
# Example input data
input_data = {
    'Brand': 'Maruti Suzuki',
    'Model_Name': 'Dzire',
    'Model_Variant': 'Base',
    'Car_Type': 'Sedan',
    'Transmission': 'Manual',
    'Fuel_Type': 'Petrol',
    'Year': 2020,
    'Kilometers': 65203,
    'Owner': '1st',
    'State': 'Bihar',
    'Accidental': 'No',

}
```

```python
# One-hot encode 'Brand'
brand_encoded = onehot_encoder_brand.transform([[input_data['Brand']]]).toarray()
brand_encoded_df = pd.DataFrame(brand_encoded, columns=onehot_encoder_brand.get_feature_names_out(['Brand']))

# One-hot encode 'Fuel_Type'
fuel_type_encoded = onehot_encoder_fuel_type.transform([[input_data['Fuel_Type']]]).toarray()
fuel_type_encoded_df = pd.DataFrame(fuel_type_encoded, columns=onehot_encoder_fuel_type.get_feature_names_out(['Fuel_Type']))


# One-hot encode 'Model_Name'
model_name_encoded = onehot_encoder_model_name.transform([[input_data['Model_Name']]]).toarray()
model_name_encoded_df = pd.DataFrame(model_name_encoded, columns=onehot_encoder_model_name.get_feature_names_out(['Model_Name']))

# One-hot encode 'Model_Variant'
model_variant_encoded = onehot_encoder_model_variant.transform([[input_data['Model_Variant']]]).toarray()
model_variant_encoded_df = pd.DataFrame(model_variant_encoded, columns=onehot_encoder_model_variant.get_feature_names_out(['Model_Variant']))

# One-hot encode 'Car_Type'
car_type_encoded = onehot_encoder_car_type.transform([[input_data['Car_Type']]]).toarray()
car_type_encoded_df = pd.DataFrame(car_type_encoded, columns=onehot_encoder_car_type.get_feature_names_out(['Car_Type']))

# One-hot encode 'State'
state_encoded = onehot_encoder_state.transform([[input_data['State']]]).toarray()
state_encoded_df = pd.DataFrame(state_encoded, columns=onehot_encoder_state.get_feature_names_out(['State']))

# One-hot encode 'Owner'
owner_encoded = onehot_encoder_owner.transform([[input_data['Owner']]]).toarray()
owner_encoded_df = pd.DataFrame(owner_encoded, columns=onehot_encoder_owner.get_feature_names_out(['Owner']))

```

```python

input_df=pd.DataFrame([input_data])
input_df
```

```python

## Encode categorical variables
input_df['Accidental']=label_encoder_accidental.transform(input_df['Accidental'])
input_df

## Encode categorical variables
input_df['Transmission']=label_encoder_transmission.transform(input_df['Transmission'])
input_df
```

```python
## concatination one hot encoded 
input_df=pd.concat([input_df.drop("Brand",axis=1),brand_encoded_df],axis=1)
input_df

## concatination one hot encoded 
input_df=pd.concat([input_df.drop("Fuel_Type",axis=1),fuel_type_encoded_df],axis=1)
input_df

## concatination one hot encoded 
input_df=pd.concat([input_df.drop("Model_Name",axis=1),model_name_encoded_df],axis=1)
input_df

## concatination one hot encoded 
input_df=pd.concat([input_df.drop("Model_Variant",axis=1),model_variant_encoded_df],axis=1)
input_df

## concatination one hot encoded 
input_df=pd.concat([input_df.drop("Car_Type",axis=1),car_type_encoded_df],axis=1)
input_df

## concatination one hot encoded 
input_df=pd.concat([input_df.drop("State",axis=1),state_encoded_df],axis=1)
input_df

## concatination one hot encoded 
input_df=pd.concat([input_df.drop("Owner",axis=1),owner_encoded_df],axis=1)
input_df

```

```python

# Load the exact feature order used during training
with open("feature_columns.pkl", "rb") as file:
    feature_columns = pickle.load(file)

# Match the training feature order
input_df = input_df.reindex(
    columns=feature_columns,
    fill_value=0
)

print("Input shape:", input_df.shape)
print("Expected features:", scaler.n_features_in_)
```

```python
## Scaling the input data
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
## PRedict Car Price
prediction = model.predict(input_scaled)

predicted_price = float(prediction[0][0])

```

```python
print("Predicted Car Price:", predicted_price)
```
